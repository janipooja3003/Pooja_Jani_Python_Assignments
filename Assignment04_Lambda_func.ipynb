Section 2: map()

```python
# Using map() and a lambda expression, calculate the square of every number.
# Using map() and a lambda expression, calculate the cube of every number.
# Using map(), add 5 to every number.
# Using map(), subtract 10 from every number.
# Using map(), multiply every number by 100.

data = [2,4,5,6,9,10,15]
square = list(map(lambda x:pow(x,2),data))
cube = list(map(lambda x: pow(x,3), data))
add = list(map(lambda x: x+5 , data))
sub = list(map(lambda x: x-10, data))
mul = list(map(lambda x: x*100, data))
print(square)
print(cube)
print(add)
print(sub)
print(mul)
```

```python
#Using map(), convert every name to uppercase.
# Using map(), convert every name to lowercase.
# Using map(), convert every name to title case.
# Using map(), calculate the length of every word.
# Using map(), extract the first character of every word.
# Using map(), extract the last character of every word

data = ["sneha","shital","kanko","haruto"]
upper = list(map(lambda x:x.upper(),data))
lower = list(map(lambda x:x.lower(),data))
tilte = list(map(lambda x:x.title(),data))
length = list(map(lambda x:len(x),data))
first_char = list(map(lambda x:x[0],data))
last_char = list(map(lambda x:x[-1],data))
print(upper)
print(lower)
print(tilte)
print(length)
print(first_char)
print(last_char)
```

```python
#Using map(), calculate 18% GST for every price.
#Using map(), calculate the final price after adding 18% GST.
#Using map(), convert every number into its absolute value.

price = [500,100,689.9876,200,745.2341]
gst = list(map(lambda x: x*0.18,price))
total = list(map(lambda x :x + (x*0.18),price))
round_gst = list(map(lambda x : round(x, 2),gst))
round_total = list(map(lambda x : round(x, 2),total))
print(f"GST = {gst}")
print(f"TOTAL = {total}")
print(f"round_gst = {round_gst}")
print(f"round_total = {round_total}")
```

Section 5: Combined Lambda, map(), filter(), and reduce()

```python
#Using filter() and map():

# Extract all even numbers.
# Calculate the square of every even number.
# Return the squared values as a list.
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
even_numbers = list(filter(lambda x: x%2 == 0,numbers))
square_even_numbers = list(map(lambda x: x**2,even_numbers))
print(square_even_numbers)
```

```python
# Using filter() and reduce():
# Extract all odd numbers.
# Calculate their total sum.

from functools import reduce
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]
odd_numbers = list(filter(lambda x: x%2 != 0,numbers))
print(odd_numbers)
sum_odd_numbers = reduce(lambda x, y: x+y, odd_numbers)
print(sum_odd_numbers)
number_greater_than_50 = list(filter(lambda x: x>50, numbers))
print(number_greater_than_50)
sum_umber_greater_than_50  =reduce(lambda x,y : x+y,number_greater_than_50)
print(sum_umber_greater_than_50)
```

```python
# Using filter() and map():
# Extract words containing more than five characters.
# Convert the selected words to uppercase.
from functools import reduce
words = ["Python", "Azure", "Ai", "developer", "Cloud", "Analytics", "spark"]
word_length = list(filter(lambda x: len(x)>= 5,words))
start_with_A = list(filter(lambda x: x.startswith("A"),words))
lower_word = list(map(lambda x: x.lower(),start_with_A))
upper_word =list(map(lambda x: x.upper(), word_length))
print(word_length)
print(upper_word)
print(start_with_A)
print(lower_word)
```

```python
# Using filter() and reduce():
# Extract palindrome words.
# Join them into one string separated by " | ".

numbers = [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]
divisible_3_5 =list(filter(lambda x :x%3 ==0 and x%5==0, numbers))
seprated_by_i = reduce(lambda x,y: x+"|"+y,map(lambda x: str(x),divisible_3_5))

print(divisible_3_5)
print(seprated_by_i)
```

```python
# Using filter(), map(), and reduce():
# Extract temperatures greater than 30°C.
# Convert the selected values into Fahrenheit.
# Calculate the average Fahrenheit temperature.

temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]

temp_greater_than30 = list(filter(lambda x:x>=30,temperatures))
print(temp_greater_than30)
degree_to_ferenhit = list(map(lambda x:x*(9/5)+32,temp_greater_than30))
print(degree_to_ferenhit)
average_ferenhit = reduce(lambda x,y: x+y,degree_to_ferenhit)/len(degree_to_ferenhit)
print(average_ferenhit)
```

```python
# Using filter() and map():

# Extract employees earning more than 60000.
# Increase their salary by 10%.
# Return their names and updated salaries.

employees = [
    {"id": 101, "name": "Rahul", "salary": 45000},
    {"id": 102, "name": "Anil", "salary": 65000},
    {"id": 103, "name": "Sneha", "salary": 72000},
    {"id": 104, "name": "Kiran", "salary": 82000},
    {"id": 105, "name": "Meena", "salary": 55000}
]
salary_greater_than_6k = list(filter(lambda x: x["salary"] >= 60000, employees))
print(salary_greater_than_6k)
ten_percent_increase = list(map(lambda x: x["salary"] * 1.1, salary_greater_than_6k))
print(ten_percent_increase)
```
