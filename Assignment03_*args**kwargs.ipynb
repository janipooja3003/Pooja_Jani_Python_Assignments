```python
#Assignment 1: Separate Positional and Keyword Arguments

def display_argument(*args, **kwargs):
  print("posititional_count:",len(args))
  print("keyword_count:", len(kwargs))
  print("positional_values:", args)
  print("keyword_values:", kwargs)

display_argument(1, 2, 3, a="hello", b="world")
```

```python
#Sum Only Numeric Positional Arguments
def calculate_numeric(*args):
  total =0
  name = ""
  print("args:", args)
  for values in args:
    if(isinstance(values, (int, float))):
      total+= values
    else:
      name += (str(values) + " ")
  print("total of values : ", total)
  print("name : ", name)

calculate_numeric(10,30,'data',80.6,'science','engineering',30)
#calculate_numeric('data','science','engineering')
```

```python
#Student Marks Report
def student_report(name,*marks,**kwargs):
  print("student_name: ", name)
  print("total_marks:", sum(marks))
  print("additional_info:", kwargs)
  student_avg = sum(marks)/len(marks)
  print("student_avg_marks:",student_avg)

  if student_avg >= 35:
    print("\n RESULT: PASS")
  else:
    print("\n RESULT: FAIL")

student_report( "kanako", 90, 80, 90, 90, course="Python", city="Hyderabad" )
student_report( "Anil", 35, 34, 19, 14, course="SQL", city="Mumbai" )
```

```python
#Count Data Types in *args
def count_data_types(*args):
  print("args:",args)
  int_count =0;
  str_count =0
  float_count = 0
  other_count = 0
  list_count = 0
  for values in args:

    if(isinstance(values , bool)):
      other_count +=1
    elif(isinstance(values, float)):
      float_count +=1
    elif(isinstance(values,str)):
      str_count+=1
    elif(isinstance(values,list)):
      list_count +=1
    elif(isinstance(values,int)):
      int_count +=1

  print("int_count:",int_count)
  print("str_count:",str_count)
  print("float_count:",float_count)
  print("list_count:",list_count)
  print("other_count:",other_count)

count_data_types(10,"abc","data",[10,20],80,50.6,90,True,"science",10.5,[60,40],9.7)
```

```python
#Filter Even and Odd Numbers
def find_even_odd(*args):
  print("vaues:",args)
  for values in args:
    if(values%2==0):
      print("even_numbers:",values)
    else:
      print("odd_numbers:",values)


find_even_odd(10,30,43,70,153,60,893,80,97,100)



```

```python
#Shopping Cart Total
def shopping_cart(*args,**kwargs):
  print("args:",args)
  print("kwargs:",kwargs)

  sub_total = sum(args)
  print("sub_total:",sub_total)
  discount_value = sub_total * (kwargs.get("Discount"))/100
  print("discount:",discount_value)
  tax_amount = sub_total * (kwargs.get("tax"))/100
  print("tax_amount:",tax_amount)
  total = sub_total - discount_value + tax_amount
  print("total:",round(total,2))

shopping_cart(100.287,200,300,400.689,500,Discount =10, tax =5)
```

```python
#Dynamic Configuration Manager
def dynemic_cofig(**kwargs):

  default = {
              "host_name" : "admin",
              "port" : 1048,
              "timeout" :10,
              "password" :"admin@123"

  }
  print("default_kwargs:",default)
  default.update(kwargs)
  print("updated_kwargs:",default)

dynemic_cofig()
print(" ")
dynemic_cofig(host_name='jani',port=2096,timeout=50,password='abc@123')
```

```python
def register_user(**kwargs):
  required = ["name", "email", "password"]
  for field in required:
    if field not in kwargs:
      print(f"{field} is missing")
      return

  print("registration successful")

register_user(name='pooja',password='123@abc')
print(" ")
register_user(name='jani',email="123@gmail.com",password='123@abc')
```

```python
#Create a Dynamic SQL SELECT Query
def create_sql_query(name,*args,**kwargs):
  print("emp_number:",name)
  print("args:",args)
  print("kwargs:",kwargs)

  query = f"SELECT * FROM {name}"
  if kwargs:
    condition =[]
    for key,values in kwargs.items():
      condition.append(f"{key} = '{values}'")
    query += " WHERE " + " AND ".join(condition)
  print("query:",query)




create_sql_query("Employee1", "rahul", 50000, department="IT", city="Bangalore" )
```
