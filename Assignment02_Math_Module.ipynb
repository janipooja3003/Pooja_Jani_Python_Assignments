```python
#Calculate compound interest using math.pow().
#pnr
principal_amount  = 8000000
number_of_period = 20
rate_of_intest = 6
print(math.pow(1+(rate_of_intest/100),number_of_period)*principal_amount)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Using existing values from previous cells
principal_amount = 8000000
rate_of_interest = 6 # as a percentage

# Define a range for the time period
time_periods = np.arange(1, 31) # From 1 to 30 years

# Calculate compound interest for each time period
future_values = [principal_amount * math.pow(1 + (rate_of_interest / 100), t) for t in time_periods]

# Calculate interest earned (Future Value - Principal Amount)
interest_earned = [fv - principal_amount for fv in future_values]

# Plotting the graph
plt.figure(figsize=(10, 6))
plt.plot(time_periods, [principal_amount] * len(time_periods), linestyle='--', color='red', label='Principal Amount')
plt.plot(time_periods, interest_earned, marker='o', label='Interest Earned')
plt.title('Principal Amount and Interest Earned Over Time')
plt.xlabel('Time Period (Years)')
plt.ylabel('Amount ($)')
plt.grid(True)
plt.legend()
plt.show()
```
